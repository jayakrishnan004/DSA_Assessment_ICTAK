```python
# Q14
student_csv = pd.read_csv("/content/drive/MyDrive/DSA - ICTAK/Assessment/Assessment 2 - october 6/numpy_pandas_assessment_datasets/student_performance.csv");
Departmentwise = student_csv.groupby("Department")["Final_Score"].agg(
    Number_of_Students="count",
    Average_Final_Score="mean",
    Maximum_Final_Score="max",
    Minimum_Final_Score="min"
)

# Sort by Average Final Score
Departmentwise = Departmentwise.sort_values("Average_Final_Score",ascending=False);

print(Departmentwise);
```

```python
# Q15
import numpy as np

final_score = student_csv["Final_Score"].to_numpy()

print("Mean:", np.nanmean(final_score))
print("Median:", np.nanmedian(final_score))
print("Standard Deviation:", np.nanstd(final_score))
print("25th Percentile:", np.nanpercentile(final_score, 25))
print("75th Percentile:", np.nanpercentile(final_score, 75))

# Overall mean
mean = np.nanmean(final_score)

# Students above the overall mean
print(student_csv[student_csv["Final_Score"] > mean])
```

```python
# Q16

import numpy as np

# A. Data Cleaning

# Identify missing values
print(student_csv.isnull().sum())

# Handle missing values
student_csv["Python_Score"] = student_csv["Python_Score"].fillna(student_csv["Python_Score"].mean());
student_csv["SQL_Score"] = student_csv["SQL_Score"].fillna(student_csv["SQL_Score"].mean());
student_csv["Pandas_Score"] = student_csv["Pandas_Score"].fillna(student_csv["Pandas_Score"].mean());
student_csv["NumPy_Score"] = student_csv["NumPy_Score"].fillna(student_csv["NumPy_Score"].mean());



# Average Score
student_csv["Average_Score"] = student_csv[["Python_Score", "SQL_Score", "Pandas_Score", "NumPy_Score"]].mean(axis=1)

# Pass Status
student_csv["Pass_Status"] = np.where(student_csv["Average_Score"] >= 50,"Pass","Fail")



# Top 5 students
print(student_csv.nlargest(5, "Average_Score"))

# Department with highest average score
print(student_csv.groupby("Department")["Average_Score"].mean().idxmax())

# Average attendance by department
print(student_csv.groupby("Department")["Attendance"].mean())

# Number of students who passed and failed
print(student_csv["Pass_Status"].value_counts())


# Summary

summary = student_csv.groupby("Department").agg(
    Total_Students=("Name", "count"),
    Average_Score=("Average_Score", "mean"),
    Average_Attendance=("Attendance", "mean"),
    Pass_Count=("Pass_Status", lambda x: (x == "Pass").sum()),
    Fail_Count=("Pass_Status", lambda x: (x == "Fail").sum())
)

print(summary)
```
