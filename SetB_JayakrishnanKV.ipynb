**Parking Fee Calculator**

```python
veh = input("Vehicle: ")
time = int(input("Hours:"))
#  Parking for less than 1 hour will be considered 1 hour.
if 0<time<1:
  time = 1

#  Identify the vehicle type.
if veh == "car":
  first_2_hrs = 50
  additional_hr = 20
elif veh =="bike":
  first_2_hrs = 30
  additional_hr = 10
elif veh == "suv":
  first_2_hrs = 70
  additional_hr = 30
else:
  print("Invalid Vehicle Type")

#  Calculate the parking fee.
if time > 2:
  fee = first_2_hrs + (time - 2)*additional_hr
  print("Parking Fee:", fee)
elif 0<time<2 :
  fee = first_2_hrs
  print("Parking Fee:", fee)
elif time<0:
  print("Invalid parking duration")
```

**Movie Theatre Seat Booking Analyzer**

```python
N = int(input("Total Shows"))

total_booked = 0
total_available = 0
high_demand_count = 0

highest_occupancy = 0
highest_show = ""

for i in range(N):
    show_name = input("Show Name :")
    total_seats = int(input("Total Seats :"))
    booked_seats = int(input("Booked Seats :"))

    available = total_seats - booked_seats
    occupancy = (booked_seats / total_seats) * 100

    if occupancy >= 80:
        category = "House Full"
        high_demand_count += 1

    elif occupancy >= 60:
        category = "High Demand"
        high_demand_count += 1

    elif occupancy >= 40:
        category = "Moderate"

    else:
        category = "Low Demand"

    print(f"{show_name} - Available: {available} - Occupancy: {occupancy:.2f}% - {category}")

    total_booked += booked_seats
    total_available += available

    if occupancy > highest_occupancy:
        highest_occupancy = occupancy
        highest_show = show_name

print("Highest Occupancy:", highest_show)
print("Total Booked Seats:", total_booked)
print("Total Available Seats:", total_available)
print("High Demand/House Full Shows:", high_demand_count)
```
